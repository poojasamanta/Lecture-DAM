```python
rules = {
    "Passengers": (0, 180),             # aircraft has 180 seats
    "Ticket_Price": (1000, 20000),      # domestic low-cost fare in INR
    "Delay_Minutes": (0, 600),          # a delay cannot be negative
    "Flight_Duration_Min": (60, 240),   # domestic sector
    "Customer_Rating": (1, 5)           # rating scale 1 to 5
}

for col, (low, high) in rules.items():
    bad = raw[(raw[col] < low) | (raw[col] > high)]
    print(col, "-> impossible values:", len(bad))

    if len(bad) > 0:
        print(bad[["Date", "Flight_ID", col]].to_string(index=False))
        print()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def show_boxplots(data, title):
    fig, axes = plt.subplots(3, 3, figsize=(13, 10))
    axes = axes.flatten()

    for i, col in enumerate(num_cols):
        sns.boxplot(y=data[col], ax=axes[i], color="skyblue")
        axes[i].set_title(col)
        axes[i].set_ylabel("")

    fig.suptitle(title, fontsize=15)
    plt.tight_layout()
    plt.show()

show_boxplots(raw, "BEFORE cleaning: boxplot of every variable")
```

RETAINING,REMOVING AND CAPPING OUTLIERS
 #Removing extreme values

```python
# Create a copy
df = raw.copy()

# Find ratings below 1 or above 5
bad_values = (df["Customer_Rating"] < 1) | (df["Customer_Rating"] > 5)

# Count unusual ratings
print("Invalid ratings:", bad_values.sum())

# Replace unusual ratings with missing values
df.loc[bad_values, "Customer_Rating"] = np.nan

# Compare missing cells
print("Missing cells before:", raw.isnull().sum().sum())

print("Missing cells after:", df.isnull().sum().sum())
```

```python
df = raw.copy()

# Check each column using its minimum and maximum allowed values
for col, (low, high) in rules.items():

    # Find values below the minimum OR above the maximum
    bad_rows = (df[col] < low) | (df[col] > high)

    # Print the column name and count of impossible values found
    print(col, "-> removed", bad_rows.sum(), "impossible value(s)")

    # Replace impossible values with NaN (missing values)
    # Note: This does not delete the entire row
    df.loc[bad_rows, col] = np.nan

    # Print a blank line to improve readability
    print()

# Count all missing cells in the original dataset
print("Blank cells before:", raw.isnull().sum().sum())

# Count all missing cells after replacing impossible values
print("Blank cells after:", df.isnull().sum().sum())
```
